### K-Means vs DBSCAN

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.cluster import DBSCAN, KMeans
import numpy as np

X, _=make_circles(n_samples=500, factor=.5, noise=.03, random_state=4)
Y, _=make_circles(n_samples=500, factor=.1, noise=.03, random_state=42)
```

```python
Z=np.vstack([X, Y])
```

```python
plt.figure(figsize=(4, 3))
plt.scatter(Z[:, 0], Z[:, 1], s=5)
plt.show()
```

## DBSCAN

```python
dbscan=DBSCAN(eps=0.1, min_samples=5)
dbscan_clusters=dbscan.fit_predict(Z)

plt.figure(figsize=(4, 3))
plt.scatter(Z[:, 0], Z[:, 1], s=5, c=dbscan_clusters, cmap='viridis')
plt.show()
```

## KMeans

```python
from sklearn.cluster import KMeans

kmeans=KMeans(n_clusters=3)
kmeans_clusters=kmeans.fit_predict(Z)

plt.figure(figsize=(4, 3))
plt.scatter(Z[:, 0], Z[:, 1], s=5, c=kmeans_clusters, cmap="viridis")
plt.show()
```

```python
# Both
fig, axs=plt.subplots(1, 2, figsize=(10, 4))

# first plot
axs[0].scatter(Z[:, 0], Z[:, 1], s=5, c=dbscan_clusters, cmap="viridis")
axs[0].set_title("DBSCAN")

# second plot
axs[1].scatter(Z[:, 0], Z[:, 1], s=5, c=kmeans_clusters, cmap="viridis")
axs[1].set_title("KMeans")

plt.tight_layout()
plt.show()
```
